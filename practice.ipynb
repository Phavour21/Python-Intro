```python
def celsius_to_fahrenheit(celsius):
    f = (celsius * 9/5) + 32
    
    return f
```

```python
def send_message(recipient, message, urgent=False):
    
    if urgent:
        status = f"[urgent] To: {recipient} - {message}"
        print(status)
    else:
        status = f"To: {recipient} - {message}"
        print(status)

send_message("samson", "the road is bad!", urgent=True)
send_message("samson", "the road is bad!")
send_message("samson", "the road is bad!")
   
```
